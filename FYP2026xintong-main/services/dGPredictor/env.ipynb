```python
# 1. 下载 Miniconda 安装包
!wget -O miniconda.sh https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh

# 2. 安装到 ~/miniconda
!bash miniconda.sh -b -p $HOME/miniconda

# 3. 查看 conda 版本（验证成功）
!$HOME/miniconda/bin/conda --version
```

```python
# 创建 dGPredictor 环境
!$HOME/miniconda/bin/conda create -n dGPredictor python=3.8.10 ipython -y

# 安装依赖
!$HOME/miniconda/bin/conda install -n dGPredictor -c conda-forge rdkit pandas matplotlib openbabel -y
!$HOME/miniconda/envs/dGPredictor/bin/pip install -U scikit-learn streamlit pulp ipykernel

# 注册到 Jupyter 内核
!$HOME/miniconda/envs/dGPredictor/bin/python -m ipykernel install --user --name dGPredictor --display-name "Python (dGPredictor)"
```
